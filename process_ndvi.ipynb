```python
#import libraries
import rasterio 
import numpy as np
```

```python
#load bands

with rasterio.open("data/final_data/B04.jp2") as red_band:
    red = red_band.read(1).astype('float32')
with rasterio.open("data/final_data/B08.jp2") as nir_band:
    nir = nir_band.read(1).astype('float32')
```

```python
#Calculate NDVI
ndvi = (nir - red) / (nir + red+ 0.0001)  # Adding a small constant to avoid division by zero
np.save("data/final_data/ndvi.npy", ndvi)
```

```python
#Save as TIFF for QGIS

transform = red_band.transform
with rasterio.open(
    "data/final_data/ndvi.tif",
    'w',
    driver='GTiff',
    height=ndvi.shape[0],
    width=ndvi.shape[1],
    count=1,
    dtype=ndvi.dtype,
    crs=red_band.crs,
    transform=transform
) as dst:
    dst.write(ndvi, 1
)
```
